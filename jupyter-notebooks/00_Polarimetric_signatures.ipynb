
## How to Use the Notebooks

To run these notebooks, follow these steps:


### 1. Install **`gdal`** Package
Installing `gdal` via `pip` may fail. Therefore, we recommend using conda to install it:
```bash
conda install gdal -c conda-forge
```

### 2. Install **`polsartools`** Package

Before running the notebooks, you'll need to install the core **`polsartools`** package. 

**Option A:** Install from PyPI `pip` (stable release)

```bash
pip install polsartools
```

**Option B:** Install from conda (stable release)

```bash
conda install polsartools -c bnarayanarao
```


**Option C:**  Install the latest version from GitHub (if you need recent updates or fixes)

```bash
pip install git+https://github.com/Narayana-Rao/polsartools.git#egg=polsartools
```
> **Note:** If you encounter a `"module not found"` error or require newer features, prefer the GitHub installation (Option C).

### 3. Verify the Installation

After successfully installing `polsartools`, you can verify it by importing the package:

```bash
import polsartools as pst
```

If this runs without errors, you're ready to explore and run the notebooks.




````python
import polsartools as pst
import numpy as np
````

### A. Polarimetric signature of an ideal trihedral corner reﬂector 
#### A.1 Default plot

````python
# 2x2 scattering matrix, S2 for an ideal trihedral
S2 = np.array([[1,0],
               [0,1]]
             )
# provide a 2x2 S2 or 3x3 Coherency matrix
pst.signature_fp(S2, 
            title='Trihedral Corner reflector', 
             # pname='',  path to the output file (*.png)
             cmap='jet', 
             plotType=1, # 1,2,3,4
                )
````

#### A.2 2D plot

````python

pst.signature_fp(S2,plotType=2)
````

#### A.3 plot on Poincaré sphere

````python
pst.signature_fp(S2,plotType=3)
````

#### A.4 Theoretical representation of polarization as a point on the Poincaré sphere

````python
pst.signature_fp(plotType=4)
````

#### A.5 Extra arguments

````python
pst.signature_fp(S2,title='trihedral corner reflector', pname='trihedral.png',cmap='viridis',plotType=1)
````

### Polarimetric signature of a trihedral corner reﬂector from airborne SAR data 


````python
S2 = np.array([[ 2248.6506   -5218.124j,      -4.0451455  -46.991398j],
               [-11.6204195  -27.771238j,      800.1831   -5830.6836j]],
    dtype=np.complex64)

pst.signature_fp(S2, title='Trihedral corner reflector',cmap='jet')
````

### Polarimetric signature of an ideal dihedral corner reﬂector 

````python
S2 = np.array([[1,0],
               [0,-1]])
pst.signature_fp(S2)
````

### Polarimetric signature of a rotated ideal dihedral corner reﬂector by 45 deg. relative to the horizontal axis


````python
S2 = np.array([[0,1],
               [1,0]])
pst.signature_fp(S2)
````
